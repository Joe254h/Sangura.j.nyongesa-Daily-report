### Exercise 11:
Write a Python program that checks if a number is prime.

```python
# Exercise 11
```

````python
# Check if a number is prime
num = 11
if num > 1:
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
````

### Exercise 48:
Write a Python function that checks if a number is prime by dividing it by numbers up to its square root.

```python
# Exercise 48
```

````python
# Check if number is prime (using sqrt)
import math

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(13))
````
